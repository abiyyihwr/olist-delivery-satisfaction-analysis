```python
import sqlite3, pandas as pd, glob, os

# 1. Tentukan alamat LENGKAP ke folder project utama kamu
# Gunakan huruf r di depan kutip agar Windows tidak bingung dengan tanda backslash (\)
folder_utama = r"C:\Project_Data\e_commerce_olist"

# 2. Buat database langsung di folder utama agar rapi
database_path = os.path.join(folder_utama, "olist.db")
conn = sqlite3.connect(database_path)

# 3. Cari folder olist_data secara otomatis di dalam folder utama
# Kita cari semua file .csv di semua sub-folder agar pasti ketemu
path_pencarian = os.path.join(folder_utama, "**", "*.csv")
file_csv = glob.glob(path_pencarian, recursive=True)

print(f"Hasil Pencarian: Menemukan {len(file_csv)} file CSV di seluruh folder proyek.")

# 4. Jika file ditemukan, masukkan ke database
if len(file_csv) > 0:
    for file in file_csv:
        nama_file_saja = os.path.basename(file)
        table_name = nama_file_saja.replace("olist_", "").replace("_dataset.csv", "")
        
        print(f" -> Memproses tabel '{table_name}' dari lokasi: {file}")
        
        df = pd.read_csv(file)
        df.to_sql(table_name, conn, if_exists="replace", index=False)
        
    print("\n[SUKSES] Semua data berhasil bermigrasi!")
else:
    print("\n[PERINGATAN] Python sama sekali tidak menemukan file .csv di C:\\Project_Data\\e_commerce_olist\\")
    print("Coba cek File Explorer Windows kamu, di manakah letak file .csv berada?")

conn.close()
```
